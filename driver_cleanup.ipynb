## Driver cleanup

Setter driver til ukjent entity der data ikke ble funnet i database.
Oppdateres etter manuell sjekk for eksistens i data.

```python
from mysql.connector import IntegrityError
from lib.data_agents import DbService
from uuid import uuid4

db_service = DbService()
```

```python
# variabler

id = str(uuid4())
driver_source_id = ""
driver_name = f"Ukjent {driver_source_id}"
driver_license_id = "06408d56-c865-462c-a4a5-5d5267e0198d"
```

```python
# commands

insert_driver = f"INSERT INTO Driver (Id, SourceId, DriverLicenseId, Name, YearOfBirth, Monte) VALUES ('{id}', {driver_source_id}, '{driver_license_id}', '{driver_name}', 0, false);"
update_buffer = f"UPDATE DriverBuffer SET Status = 1 WHERE SourceId = {driver_source_id};"

check_buffer = f"SELECT * FROM DriverBuffer WHERE SourceId = {driver_source_id};"
check_driver = f"SELECT * FROM Driver WHERE SourceId = '{driver_source_id}';"
```

```python
# check source id here:
if driver_source_id == "":
  raise Exception("Driver source id is empty")

test = db_service.query(check_buffer)
if len(test) == 0:
  raise Exception("Driver source id not found")
```

```python
# insert driver and get result
try:
    db_service.command(insert_driver)
except IntegrityError:
  print(f"Driver with SourceId '{driver_source_id}' already exists!!!")

driver = db_service.query(check_driver)[0]
print(f"Driver source id: '{driver[1]}'")
print(f"Driver name ....: '{driver[3]}'")
```

```python
# update and show buffer
db_service.command(update_buffer)

buffer = db_service.query(check_buffer)[0]
print(f"Driver source id: '{buffer[1]}'")
print(f"Driver status ..: '{buffer[2]}'")
```
