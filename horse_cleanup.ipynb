## Horse Cleanup

Setter horse til ukjent entity med source id der data ikke ble funnet i databasen.

Oppdateres etter manuell sjekk for eksistens i data!

```python
from multiprocessing.managers import rebuild_as_list

from mysql.connector import IntegrityError
from lib.data_agents import DbService
from uuid import uuid4

db_service = DbService()
```

```python
# variabler

id = str(uuid4())
horse_source_id = ""
horse_name = f"Ukjent {horse_source_id}"
horse_sex_id = "6cfe5750-ed43-4fa4-b19c-1adabf9a8266"
horse_type_id = "9571b064-78d9-453a-b38b-d76d063da985"
```

```python
# commands and queries

insert_horse = f"INSERT INTO Horse (Id, SourceId, HorseSexId, HorseTypeId, Name, YearOfBirth) VALUES ('{id}', '{horse_source_id}', '{horse_sex_id}', '{horse_type_id}', '{horse_name}', 0);"
update_buffer = f"UPDATE HorseBuffer SET Status = 1 WHERE SourceId = '{horse_source_id}';"

check_buffer = f"SELECT * FROM HorseBuffer WHERE SourceId = '{horse_source_id}';"
check_horse = f"SELECT * FROM Horse WHERE SourceId = '{horse_source_id}';"
```

```python
if horse_source_id == "":
  raise Exception("Horse Source ID cannot be empty!")

test = db_service.query(check_buffer)
if len(test) == 0:
  raise Exception("No horses found")
```

```python
try:
  db_service.command(insert_horse)
except IntegrityError:
  print(f"Horse with SourceId '{horse_source_id}' already exists!!!")

horse = db_service.query(check_horse)[0]
print(f"Horse source id: {horse[1]}")
print(f"Horse name: {horse[4]}")
```

```python
db_service.command(update_buffer)

buffer = db_service.query(check_buffer)[0]
print(f"Horse source id: {buffer[1]}")
print(f"Horse status: {buffer[2]}")
```
